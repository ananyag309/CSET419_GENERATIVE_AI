In this experiment, we implemented a Bigram language model, which is a statistical N-gram model. It predicts the next word based only on the previous word by storing word-to-word mappings in a dictionary. During generation, the next word is randomly selected from possible successors.

```python
from collections import defaultdict
import random
import string
```

```python
with open("sample text.txt", "r", encoding="utf-8") as f:
    text = f.read()
```

```python
# Convert to lowercase and remove punctuation
text = text.lower().translate(
str.maketrans("", "", string.punctuation)
)
# Split into words
words = text.split()
```

```python
#Bigram Model

bigram_model = defaultdict(list)
for i in range(len(words) - 1):
    bigram_model[words[i]].append(words[i + 1])
```

```python
def generate_bigram(seed, length=30):
    result = [seed]
    for _ in range(length):
        next_words = bigram_model.get(result[-1])
        if not next_words:
            break
        result.append(random.choice(next_words))
    return " ".join(result)
```

```python
seed = random.choice(words)
print("Generated Text (Bigram):\n")
print(generate_bigram(seed))
```
